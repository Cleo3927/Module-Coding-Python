### 1. Data & Context
For this test submission, I've chosen a sample text about artificial intelligence and technology. This text contains various patterns like emails, dates, URLs, and technical terms that will help demonstrate regex extraction and NLP preprocessing capabilities. The text is approximately 400 words and covers current trends in AI development.

```python
# Sample text data
sample_text = """
Artificial Intelligence Revolution: A Modern Perspective

The field of artificial intelligence has experienced tremendous growth since 2020. 
Major tech companies like Microsoft, Google, and OpenAI have been at the forefront 
of this revolution. For more information, contact us at info@aitech.com or visit 
https://www.aitech.com for the latest updates.

On January 15, 2023, ChatGPT reached 100 million users, setting a new record. 
By March 1, 2024, the number had doubled. Research papers published on October 20, 2024, 
show that large language models (LLMs) now have capabilities that were unimaginable 
just five years ago.

Machine learning algorithms process millions of data points daily. Natural language 
processing (NLP) enables computers to understand human language with unprecedented 
accuracy. Deep learning networks, particularly transformers, have revolutionized 
how we approach complex problems.

The technology sector invested $50 billion in AI research in 2023 alone. Companies 
are hiring data scientists at salaries ranging from $120,000 to $250,000 annually. 
Contact our recruitment team at careers@techcorp.com or call +1-555-123-4567 for 
opportunities.

Key developments include:
- Neural networks achieving 95% accuracy in image recognition
- Processing speeds improving by 300% since 2020
- 10,000+ research papers published on AI topics in 2024
- $25.5 million funding rounds for AI startups

Educational institutions have adapted their curricula to include AI and machine learning 
courses. Students are learning Python, TensorFlow, and PyTorch to build intelligent 
systems. The future of technology depends on continuous innovation and ethical AI 
development.

For technical support, email support@aitech.com or visit our documentation at 
https://docs.aitech.com/api. The AI community continues to grow, with forums at 
https://community.aitech.com hosting discussions on breakthrough technologies.
"""

print("Text loaded successfully!")
print(f"Text length: {len(sample_text)} characters")
print(f"Word count: {len(sample_text.split())} words")
```

### 2. Regex Extraction
I'll implement regex patterns to extract emails, dates, phone numbers, and dollar amounts.

```python
import re

# Pattern 1: Email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, sample_text)
print("=== Email Addresses ===")
print(f"Found {len(emails)} emails:")
for email in emails:
    print(f"  - {email}")
print("Accuracy note: This pattern handles standard email formats but may not catch all edge cases like special characters in local parts.\n")

# Pattern 2: Dates (Month Day, Year format)
date_pattern = r'\b(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{1,2},\s+\d{4}\b'
dates = re.findall(date_pattern, sample_text)
print("=== Dates ===")
print(f"Found {len(dates)} dates:")
for date in dates:
    print(f"  - {date}")
print("Accuracy note: This pattern specifically matches 'Month Day, Year' format. Other date formats would require additional patterns.\n")

# Pattern 3: Phone numbers (US format)
# Pattern handles formats: +1-555-123-4567, (555) 123-4567, 555-123-4567
phone_pattern = r'\+?\d{1,3}[-.]?(?:\(\d{3}\)|\d{3})[-.]?\d{3}[-.]?\d{4}\b'
phones = re.findall(phone_pattern, sample_text)
print("=== Phone Numbers ===")
print(f"Found {len(phones)} phone numbers:")
for phone in phones:
    print(f"  - {phone}")
print("Edge case: This pattern handles various formats like +1-555-123-4567, (555) 123-4567, etc.\n")

# Pattern 4: Dollar amounts
money_pattern = r'\$[\d,]+(?:\.\d{2})?(?:\s+(?:billion|million))?'
amounts = re.findall(money_pattern, sample_text)
print("=== Dollar Amounts ===")
print(f"Found {len(amounts)} monetary values:")
for amount in amounts:
    print(f"  - {amount}")
print("Accuracy note: Captures amounts with optional million/billion qualifiers.")
```

### 3. NLP Preprocessing
I'll use tokenization, lowercase conversion, stop word removal, and lemmatization (chosen for better word normalization than stemming).

```python
# Install required packages if needed
# !pip install nltk

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import Counter

# Download required NLTK data
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

print("=== NLP Preprocessing Pipeline ===")

# Step 1: Tokenization
tokens = word_tokenize(sample_text)
print(f"1. Tokenization: {len(tokens)} tokens")

# Step 2: Lowercase
tokens_lower = [token.lower() for token in tokens]
print(f"2. Lowercase: {len(tokens_lower)} tokens")

# Step 3: Remove stop words and non-alphabetic tokens
stop_words = set(stopwords.words('english'))
tokens_filtered = [token for token in tokens_lower if token.isalpha() and token not in stop_words]
print(f"3. Stop word removal: {len(tokens_filtered)} tokens remaining")

# Step 4: Lemmatization
# Why lemmatization? It produces real words (e.g., 'processing' -> 'process') 
# whereas stemming can produce non-words (e.g., 'processing' -> 'process' but 'studies' -> 'studi')
lemmatizer = WordNetLemmatizer()
tokens_lemmatized = [lemmatizer.lemmatize(token) for token in tokens_filtered]
print(f"4. Lemmatization complete")
print(f"\nWhy lemmatization? It provides proper word forms, making results more interpretable.")

# Count and display top 15 tokens
token_counts = Counter(tokens_lemmatized)
top_15_tokens = token_counts.most_common(15)

print("\n=== Top 15 Tokens After Preprocessing ===")
for i, (token, count) in enumerate(top_15_tokens, 1):
    print(f"{i:2d}. {token:20s} - {count} occurrences")
```

### 4. Regex + NLP Combo
Extract number-noun pairs to identify quantified entities in the text.

```python
import nltk

# Download POS tagger if needed
nltk.download('averaged_perceptron_tagger', quiet=True)

print("=== Regex + NLP Combo: Number-Noun Pairs ===")
print("Extracting quantified entities from the text...\n")

# Find patterns like "100 million users", "95% accuracy", etc.
# Pattern matches: number (with optional decimal/thousands separator and % sign) followed by word(s)
number_context_pattern = r'\b(\d+(?:[.,]\d+)?%?)\s+(\w+(?:\s+\w+){0,2})'
matches = re.findall(number_context_pattern, sample_text)

print("Found number-entity pairs:")
for i, (number, context) in enumerate(matches[:10], 1):  # Show first 10
    # Simple POS tagging to verify we have nouns
    tokens = word_tokenize(context)
    pos_tags = nltk.pos_tag(tokens)
    has_noun = any(tag.startswith('NN') for _, tag in pos_tags)
    
    if has_noun:
        print(f"  {i}. {number:15s} → {context}")

print("\nComment: This combines regex number extraction with NLP to identify")
print("quantified entities, useful for extracting statistics and measurements.")
```

### 5. Visualization
Create a bar chart showing the top 15 most common tokens.

```python
import matplotlib.pyplot as plt

# Prepare data for visualization
tokens_list = [token for token, count in top_15_tokens]
counts_list = [count for token, count in top_15_tokens]

# Create bar chart
plt.figure(figsize=(12, 6))
plt.bar(tokens_list, counts_list, color='steelblue', edgecolor='navy', alpha=0.7)
plt.xlabel('Tokens', fontsize=12, fontweight='bold')
plt.ylabel('Frequency', fontsize=12, fontweight='bold')
plt.title('Top 15 Most Common Tokens After NLP Preprocessing', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()

print("\nVisualization shows that 'ai', 'technology', 'learning', and 'intelligence'")
print("are the most frequent meaningful terms, reflecting the AI-focused content.")
```
