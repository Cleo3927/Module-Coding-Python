```python
# French text
french_text = """Qu’est-ce que Labubu exactement ?
C’est une question qui continue de troubler bien des gens – et même ceux qui connaissent la réponse ne sont pas tout à fait sûrs de pouvoir expliquer cette folie.
Labubu est à la fois un personnage fictif et une marque. Le mot lui-même ne signifie rien. C’est le nom d’un personnage issu de la série de jouets « Les Monstres », créée par l’artiste hongkongais Kasing Lung.
Les visages en vinyle sont fixés à des corps en peluche et possèdent un aspect reconnaissable : des oreilles pointues, de grands yeux et un sourire malicieux montrant exactement neuf dents. Internet, à la fois intrigué et divisé, ne parvient pas à se mettre d’accord pour savoir s’ils sont adorables ou étranges.
Selon le site officiel du distributeur, Labubu est « gentil et veut toujours aider, mais obtient souvent le résultat contraire par accident. »
Les poupées Labubu sont apparues dans plusieurs séries de « Les Monstres », telles que « Plein d’énergie », « Prends une place », « Macaron excitant » et « Tombe dans la nature ».
La marque Labubu comprend aussi d’autres personnages issus du même univers, qui ont inspiré leurs propres poupées populaires – comme la cheffe du clan Zimomo, son petit ami Tycoco et leur amie Mokoko.

Qui vend Labubu ?
Une grande partie des ventes de Pop Mart provenait, pendant plusieurs années, des boîtes surprises – où les clients découvraient ce qu’ils avaient acheté seulement après avoir ouvert l’emballage. C’est durant cette période que Pop Mart a collaboré avec Kasing Lung pour obtenir les droits sur Labubu.
C’était en 2019, presque dix ans après que l’entrepreneur Wang Ning eut fondé Pop Mart à Pékin, à l’origine comme un magasin de type bazar. Après le succès des boîtes surprises, Pop Mart lança sa première série en 2016, en vendant les poupées Molly, des figurines enfantines créées par l’artiste hongkongais Kenny Wong.
"""

# Tokenization
enc = tiktoken.encoding_for_model("gpt-3.5-turbo")
french_tokens = enc.encode(french_text)
french_token_pieces = [enc.decode([t]) for t in french_tokens]

print("Total tokens:", len(french_tokens))
print("French tokens:")
print(french_token_pieces)
```

```python
# Define a function that computes word-level token statistics

def word_level_stats_from_tokens(text, enc, token_pieces):

    # Split text into words (simple whitespace split)
    words = text.split()

    token_index = 0
    token_counts = []

    # Go through each word and count how many tokens correspond to it
    for word in words:
        count = 0
        word_len = len(word)
        current = ''

        # Match token pieces to approximate the word boundaries
        while token_index < len(token_pieces) and len(current.replace(' ', '')) < word_len:
            current += token_pieces[token_index]
            count += 1
            token_index += 1
        token_counts.append(count)

    # Basic totals
    total_words = len(words)
    total_tokens = len(token_pieces)  # Use the full-text token count for consistency

    # Calculate the statistics
    stats = {
        "1_token_%": round(sum(1 for c in token_counts if c == 1) / total_words * 100, 2),
        "2_tokens_%": round(sum(1 for c in token_counts if c == 2) / total_words * 100, 2),
        "3+_tokens_%": round(sum(1 for c in token_counts if c >= 3) / total_words * 100, 2),
        "avg_tokens_per_word": round(total_tokens / total_words, 2),
        "total_words": total_words,
        "total_tokens": total_tokens
    }

    return stats


# Calculate stats for all three languages

stats_english = word_level_stats_from_tokens(english_text, enc, english_token_pieces)
stats_german  = word_level_stats_from_tokens(german_text, enc, german_token_pieces)
stats_french  = word_level_stats_from_tokens(french_text, enc, french_token_pieces)

# Combine the results into a single comparison table
df_stats = pd.DataFrame({
    "English": stats_english,
    "German": stats_german,
    "French": stats_french
}).T  # transpose so that languages appear as rows

df_stats
```

### **Mini Write-up**

Based on the tokenization results, English turned out to be the most “token-efficient” language, it required the fewest tokens per word on average.
French was slightly more expensive in tokens, while German produced the highest number of tokens relative to its word count.
This shows a clear pattern: the more complex and morphologically rich the language, the higher the token count.

- **English:**
  English words are short, often one token each, and the tokenizer handles them efficiently. The language has limited inflection and very few compound forms, which keeps the tokenization compact and simple.

- **French:**
  French sits in the middle. Its words often contain inflectional endings and accented letters (like *boîtes*, *personnages*), which sometimes get split into multiple tokens.
  This makes French moderately more expensive in tokens than English, but still relatively efficient.

- **German:**
  German showed the highest tokenization cost. This happens because German frequently uses compound words (like *Überraschungsboxen* or *Vinylgesichter*) and rich morphology.
  The tokenizer tends to break such words into multiple meaningful subparts, so the average tokens per word are noticeably higher.

### **Graph conclusions**

  This visually demonstrates how linguistic structure influences tokenization efficiency.

- **Stacked bar chart (Distribution of 1-, 2-, and 3+ Token Words):**
  English has the largest share of single-token words, while German has the smallest.
  French again appears balanced, showing both one- and two-token words in moderate proportions.

- **Scatter plot (Words vs. Tokens):**
  The scatter plot (if included) reinforces this trend - English points lie close to the diagonal where words ~ tokens, French lies slightly above, and German sits far above it, meaning that German needs many more tokens to represent the same number of words.


Overall, English is the most token-efficient language, German the least, and French falls somewhere in between.
The reason lies in the languages’ structures: English’s simpler morphology makes it cheaper for the tokenizer, while German’s long compounds and inflectional endings make it more “expensive.”
These results illustrate how linguistic complexity directly impacts how a model represents text in tokens.
